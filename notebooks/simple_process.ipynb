```python
from openai import OpenAI
import httpx
```

```python
import os
client = OpenAI(
        base_url="https://openrouter.ai/api/v1",
        api_key=os.environ["OPENROUTER_API_KEY"],
        http_client=httpx.Client(proxy="http://127.0.0.1:10808"),
    )
```

```python
import json
def getTopic(text):
    model = "nvidia/nemotron-3.5-lightning:free"
    response = client.chat.completions.create(
        model=model,
        messages=[
            {
                "role": "system", 
                "content": (
                    "موضوع متن فارسی را استخراج کن و نتیجه را فقط به صورت یک شیء JSON "
                    "با کلید 'topic' برگردان. هیچ توضیح اضافی، جمله یا کاراکتر دیگری خارج از JSON ننویس."
                )
            },
            {
                "role": "user", 
                "content": f"متن: {text}"
            }
        ],
        response_format={"type": "json_object"}
    )
    content = response.choices[0].message.content
    try:
        topic_data = json.loads(content)
        topic = topic_data.get("topic")
        return topic
    except json.JSONDecodeError:
        print("خروجی مدل JSON معتبر نبود:", content)
        return "سایر"
```

```python
untiteldRows["topic"] = untiteldRows["generated_text"].apply(getTopic)
```
